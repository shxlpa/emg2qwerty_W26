```python
import pandas as pd
from tensorboard.backend.event_processing.event_accumulator import EventAccumulator
from pathlib import Path

base = Path("./logs/default-hybrid-CNN-LSTM/")

best_rows = []

for pct in pct_values:
    print(f"Processing logs for training size: {pct}%")

    training_dir = next(base.glob(f"Training_{pct}"))

    # keep the checkpoint identification portion
    ckpts = sorted(training_dir.glob("**/epoch*.ckpt"))
    checkpoint = ckpts[-1] if ckpts else None

    log_dir = next(training_dir.glob("**/lightning_logs/version_0"))
    event_files = list(log_dir.glob("events.out.tfevents*"))

    epoches = []
    val_cers = []

    for event_file in event_files:
        event_acc = EventAccumulator(str(event_file))
        event_acc.Reload()

        tags = event_acc.Tags()["scalars"]
        if "val/CER" in tags and "epoch" in tags:
            val_cer_events = event_acc.Scalars("val/CER")
            epoch_events = event_acc.Scalars("epoch")

            epoch_map = {e.step: int(e.value) for e in epoch_events}

            for event in val_cer_events:
                if event.step in epoch_map:
                    epoches.append(epoch_map[event.step])
                    val_cers.append(event.value)

    df_metrics = pd.DataFrame({
        "epoch": epoches,
        "val/CER": val_cers
    })

    df_metrics = df_metrics.drop_duplicates(subset=["epoch"], keep="last")
    df_metrics = df_metrics.sort_values(by="epoch").reset_index(drop=True)

    # one point only: best CER for this % of data
    best_idx = df_metrics["val/CER"].idxmin()
    best_row = df_metrics.loc[best_idx].copy()

    best_row["training_data_size"] = pct
    best_row["checkpoint_name"] = checkpoint.name if checkpoint else None
    best_row["checkpoint_path"] = str(checkpoint) if checkpoint else None

    best_rows.append(best_row)

df_best_cer = pd.DataFrame(best_rows).sort_values("training_data_size").reset_index(drop=True)

print(df_best_cer)
```

```python
base = Path("./logs/default-LSTMEncoder/")

test_metrics_all = []
test_rows = []

for pct in pct_values:
    print(f"Processing test logs for training size: {pct}%")

    training_dir = next(base.glob(f"Training_{pct}"))
    checkpoint = sorted(training_dir.glob("**/epoch*.ckpt"))[-1]

    testing_dir = next(base.glob(f"Testing_{pct}"))
    log_dir = next(testing_dir.glob("**/lightning_logs/version_0"))
    event_files = list(log_dir.glob("events.out.tfevents*"))

    epoches = []
    test_cers = []

    for event_file in event_files:
        event_acc = EventAccumulator(str(event_file))
        event_acc.Reload()

        tags = event_acc.Tags()["scalars"]

        if "test/CER" in tags and "epoch" in tags:
            test_cer_events = event_acc.Scalars("test/CER")
            epoch_events = event_acc.Scalars("epoch")
            epoch_map = {e.step: int(e.value) for e in epoch_events}

            for event in test_cer_events:
                if event.step in epoch_map:
                    epoches.append(epoch_map[event.step])
                    test_cers.append(event.value)

    df_metrics = pd.DataFrame({
        "epoch": epoches,
        "test/CER": test_cers
    })

    df_metrics = df_metrics.drop_duplicates(subset=["epoch"], keep="last")
    df_metrics = df_metrics.sort_values(by="epoch").reset_index(drop=True)

    if len(df_metrics) == 0:
        continue

    df_metrics["training_data_size"] = pct
    df_metrics["checkpoint_name"] = checkpoint.name
    df_metrics["checkpoint_path"] = str(checkpoint)

    # keep all test points
    test_metrics_all.append(df_metrics)

    # also keep one summary row if you still want it
    row = df_metrics.iloc[-1].copy()
    test_rows.append(row)

df_test_all = pd.concat(test_metrics_all, ignore_index=True)
df_test_cer = pd.DataFrame(test_rows).sort_values("training_data_size").reset_index(drop=True)
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Keep only the columns needed for plotting
df_val_plot = df_best_cer[["training_data_size", "val/CER"]].copy()
df_test_plot = df_test_cer[["training_data_size", "test/CER"]].copy()

# Merge validation and test CER by training data size
df_plot = pd.merge(df_val_plot, df_test_plot, on="training_data_size", how="inner")

# Print best val and test CER for each training size
print("Best Validation and Test CER by Training Data Size:\n")
for _, row in df_plot.sort_values("training_data_size").iterrows():
    print(
        f"{int(row['training_data_size'])}% data -> "
        f"Best Val CER: {row['val/CER']:.4f}, "
        f"Test CER: {row['test/CER']:.4f}"
    )

# Convert to long format for seaborn
df_plot_long = df_plot.melt(
    id_vars="training_data_size",
    value_vars=["val/CER", "test/CER"],
    var_name="metric",
    value_name="CER"
)

plt.figure(figsize=(12, 7))
sns.lineplot(
    data=df_plot_long,
    x="training_data_size",
    y="CER",
    hue="metric",
    marker="o"
)

plt.title("Validation and Test CER vs Training Data Size")
plt.xlabel("Training Data Size (%)")
plt.ylabel("CER")
plt.grid(True)
plt.xticks(pct_values)
plt.legend(title="Metric")
plt.tight_layout()
plt.show()
```
